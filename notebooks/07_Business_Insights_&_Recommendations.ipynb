## Executive Summary

The analysis of the Olist e-commerce marketplace revealed strong revenue concentration alongside a significant customer retention challenge.

Key findings include:

- Total revenue reached approximately **13.22 million**
- **96,478** delivered orders were analyzed
- Approximately **96,096** customers were served
- Average order value was approximately **137.04**
- The top 10 product categories generated **63.22%** of revenue
- The top 3 states generated **63.38%** of revenue
- The top 10% of customers generated **41.10%** of total revenue
- VIP customers accounted for **62.53%** of historical customer value
- Only about **3%** of customers made repeat purchases
- Monthly cohort retention generally fell below **1%** after the initial purchase
- VIP customers showed the highest repeat-purchase rate at **7.19%**
- Predictive modeling showed that **recency** was the strongest indicator of future repeat purchasing

Overall, the marketplace generates substantial transaction volume but has a highly transactional customer base with limited repeat purchasing.

## Customer Value & Segmentation Insights

Customer value is highly concentrated among a relatively small group of customers.

The VIP CLV segment represents approximately one-quarter of customers but contributes **62.53% of historical customer value**. Similarly, the top 10% of customers account for **41.10% of total revenue**.

RFM analysis identified Loyal Customers and Potential Loyalists as two of the largest strategically valuable segments, while At Risk and Hibernating customers also represent substantial revenue that could potentially be recovered through targeted engagement.

These findings indicate that customer retention strategies should prioritize high-value and high-potential segments rather than applying the same approach to all customers.

## Retention Insights

Customer retention is the primary weakness identified in the marketplace.

Only about **3%** of customers made more than one purchase. Cohort analysis also showed that monthly retention generally falls below **1%** after the initial purchase, demonstrating that the low repeat-purchase behavior persists across customer acquisition periods.

Retention improves with customer value:

- VIP: **7.19%**
- High: **2.85%**
- Medium: **1.53%**
- Low: **0.43%**

Although VIP customers are substantially more likely to return than lower-value customers, retention remains low across every segment.

## Predictive Analytics Insights

Machine learning models were developed to identify customers likely to make a future repeat purchase.

The expanded Logistic Regression model provided the strongest overall performance with:

- ROC-AUC: **0.583**
- PR-AUC: **0.0247**
- Repeat-customer recall: **53.6%**

Predictive performance remained limited because repeat purchasing represented only **0.56%** of the modeling population during the prediction window.

Recency was the strongest predictor of repeat purchasing, followed by customer tenure. Customers who purchased more recently and had a longer relationship with the marketplace were more likely to return.

Given the low precision of the model, prediction scores are more suitable for customer prioritization than automatic classification.

## Strategic Recommendations

1. **Prioritize retention over acquisition-only growth**  
   The extremely low repeat-purchase rate indicates that improving customer retention represents one of the largest opportunities for long-term growth.

2. **Protect high-value customers**  
   VIP and Loyal customers contribute a disproportionate share of revenue and should receive targeted retention incentives, personalized offers, and loyalty benefits.

3. **Encourage a second purchase quickly**  
   Because recency strongly predicts future purchasing, post-purchase campaigns should engage customers soon after their first transaction.

4. **Re-engage valuable inactive customers**  
   At Risk, Cannot Lose Them, and Hibernating customers represent meaningful historical revenue and should be targeted with win-back campaigns.

5. **Use customer propensity scores for campaign prioritization**  
   Predictive model scores can help rank customers for retention campaigns, but they should be combined with segmentation and business rules rather than used as standalone decisions.

## Conclusion

Customer segmentation, CLV analysis, retention analysis, and predictive modeling consistently point to the same strategic opportunity: improving customer retention, particularly among high-value and recently active customers.
